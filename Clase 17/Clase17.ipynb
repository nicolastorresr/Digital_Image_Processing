# Representación de objetos

- Típicamente, después de segmentar una imagen viene una etapa llamada *Feature Extraction* (extracción de características), que consiste en detectar características y describir esas características.

- Detectar características permite representar un objeto asignando atributos cuantitativos a las características detectadas.

- Por ejemplo, podríamos detectar esquinas en el borde de una región y describir esas esquinas por su orientación y ubicación, los cuales son atributos cuantitativos.

# Representación de Contornos

Representación de contorno es variada:
- Los píxeles del borde.
- Puntos representativos.
- Puntos representativos parametrizados (e.g. splines (función definida por partes mediante polinomios)).

Existen métodos que consideran un conjunto fijo de puntos y otros son más flexibles.

La dinámica de seguimiento de contornos también es variada (a estudiar más adelante).

```python
# Subdividir polígono usando Splines de segundo grado
new_hand = hand.copy()
for _ in range(5):
    new_hand = subdivide_polygon(new_hand, degree=2, preserve_ends=True)
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 4))
ax1.plot(hand[:, 0], hand[:, 1], c="black")
ax1.set_title('Original')
ax2.plot(new_hand[:, 0], new_hand[:, 1], c='green')
ax2.set_title("Splines")
ax3.plot(appr_hand[:, 0], appr_hand[:, 1], c='blue')
ax3.set_title("Douglas-Peucker")
plt.show()
```

# Active Contour Model (Snakes)

- El "Modelo de Contorno Activo" es un método para ajustar *splines* a bordes en una imagen.

- Funciona minimizando una Energía que está definida en parte por la imagen y en parte por la forma del *spline*.

- El problema de encontrar el contorno se modela como un problema de minimización de energía.

# Framework for snakes

- Se inicializa cualquier curva cercana al borde de la imagen.
- La *snake* empieza a deformarse y se mueve hacia el borde deseado.
- Al final, la serpiente se envuelve alrededor del objeto.

# Parametrización del Contorno

El contorno se define en el plano $x,y$ de una imagen, como una curva paramétrica: 

$$\mathbf{v}(s) = (x(s),y(s))$$

# Función de Energía

El contorno de la imagen posee energías que de alguna forma ejercen cierta fuerza de acción sobre él. La Energía $E_{snake}$ se define como la suma de la energía interna y la externa de la imagen:

$$ E_{snake} = E_{interna} + E_{externa} $$

- Energía interna: Fomenta preferencias de forma previas. Por ejemplo, suavidad, elasticidad, forma particular conocida.

- Energía externa (energía de “imagen”): Anima al contorno a encajar en lugares donde existen estructuras de imagen, por ejemplo, bordes.

Un buen ajuste entre el contorno deformable actual y la forma del objetivo en la imagen producirá una energía baja.

# Energía Interna

- Es la suma entre "elasticidad" y "rigidez".
- Elasticidad permite que la energía se pueda contraer.

# Ejemplo Active Contour Model (Snakes) en Python

```python
from skimage import data
from skimage.color import rgb2gray
from skimage.filters import gaussian
from skimage.segmentation import active_contour
```

```python
astronaut = data.astronaut()
astronaut = rgb2gray(astronaut)
```

```python
plt.imshow(astronaut, cmap=plt.cm.gray)
plt.show()
```

```python
s = np.linspace(0, 2*np.pi, 400)
r = 100 + 100*np.sin(s)
c = 220 + 100*np.cos(s)
region = np.array([r, c]).T
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(astronaut, cmap=plt.cm.gray)
ax.plot(region[:, 1], region[:, 0], '--b', lw=3)
ax.set_xticks([]), ax.set_yticks([])
ax.axis([0, astronaut.shape[1], astronaut.shape[0], 0])
plt.show()
```

```python
def plot_snakes(img, region, sigma, alpha, beta, gamma):
    snake = active_contour(gaussian(img, sigma, preserve_range=False), region, alpha=alpha, beta=beta, gamma=gamma)
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.imshow(img, cmap=plt.cm.gray)
    ax.plot(region[:, 1], region[:, 0], '--b', lw=3)
    ax.plot(snake[:, 1], snake[:, 0], '-r', lw=3)
    ax.set_xticks([]), ax.set_yticks([])
    ax.axis([0, img.shape[1], img.shape[0], 0])
    plt.show()
```

```python
plot_snakes(astronaut, region, sigma=3, alpha=0.015, beta=10, gamma=0.001)
```

# Morphological Snakes

*Morphological Snakes* son métodos similares a los *Active Contours*, pero usan operadores morfológicos (como dilatación o erosión) sobre la imagen binaria.

# Chan-Vese

Active Contour Model implementado con operadores morfológicos mediante la evolución de un conjunto de niveles. Se puede utilizar para objetos sin bordes claramente definidos. Se requiere que el interior del objeto se vea diferente en promedio que el exterior (es decir, el área interior del objeto debe ser más oscura o más clara que el área exterior en promedio).

```python
from skimage import data, img_as_float
from skimage.segmentation import morphological_chan_vese
```

```python
camera = img_as_float(data.camera())
```

```python
plt.imshow(camera, cmap=plt.cm.gray)
plt.show()
```

```python
def plot_chan_vese(img, num_iter, smoothing):
    ls = morphological_chan_vese(img, num_iter=num_iter, smoothing=smoothing)
    fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(16, 8), sharex=True, sharey=True)
    ax1.imshow(ls, cmap=plt.cm.gray)
    ax1.set_title('Chan-Vese')
    ax1.axis('off')
    ax2.imshow(img, cmap="gray")
    ax2.contour(ls, colors='r')
    ax2.axis('off')
    ax2.set_title('Contorno')
```

```python
plot_chan_vese(camera, num_iter=35, smoothing=3)
```
