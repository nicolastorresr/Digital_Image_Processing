# Imagen digital

- Es la Representación numérica (generalmente binaria) de una imagen bidimensional.
- Dependiendo de si la resolución de la imagen es estática o dinámica, puede tratarse de una imagen matricial (mapa de bits) o de un gráfico vectorial.

# Resolución de imagen

- La resolución de una imagen indica la cantidad de detalles que pueden observarse en esta. 

- Tener mayor resolución se traduce en obtener una imagen con más detalle o calidad visual.

- Para las imágenes digitales almacenadas como mapa de bits, la resolución se define con la cantidad de píxeles a lo ancho (columnas) y a lo largo (filas).

# Gráfico vectorial

Es una imagen digital formada por objetos geométricos dependientes (segmentos, polígonos, arcos, muros, etc.), cada uno de ellos definido por atributos matemáticos de forma, de posición, etc.

<div align="center">
    <img src="img/vector.jpg" alt="" width="45%" align="left" />
    <img src="img/CorelDRAW.png" alt="" width="40%" align="center" />
</div>


# Ventajas

- No pierden calidad al ser redimensionadas.
- Los objetos definidos por vectores pueden ser guardados y modificados en el futuro.
- Permiten mover, estirar y retorcer imágenes de manera relativamente sencilla.

# Desventajas

- Los datos que describen el gráfico vectorial deben ser procesados, es decir, el computador debe ser suficientemente potente para realizar los cálculos necesarios para formar la imagen final.
- Los gráficos vectoriales, en general, no son aptos para codificar fotografías o vídeos tomados en el «mundo real».

# Mapa de bits (ráster)

- Es una matriz de puntos (Dot Matrix en inglés), la cual consiste básicamente en una tabla donde cada elemento es un píxel (punto de color).

- Para el computador, cada píxel no es más que una tupla que se transforma en información visual.

    <img src="img/raster.jpg" alt="" width="20%" align="right" />
</div>

# Conversión entre mapas de bits y gráficos vectoriales

- La transformación de un mapa de bits a un formato vectorial se llama vectorización. Existe software dedicado al proceso como Corel PowerTrace o Inkscape.
- El proceso inverso, convertir una imagen vectorial en una imagen de mapa de bits, es mucho más sencillo y se llama rasterización.

# Síntesis aditiva

- Obtiene un color de luz determinado por la suma de otros colores.
- Normalmente utiliza luz roja, verde y azul para producir el resto de los colores.
- Definen el modelo RGB.

<div align="center">
    <img src="img/AdditiveColor.png" alt="" width="40%" align="center" />
</div>


# Síntesis sustractiva

- Los pigmentos al mezclarse, sustraen o absorben más luz.
- Los colores primarios son el amarillo, magenta y cian, cada uno de los cuales tiene la misión de absorber el campo de radiación de cada tipo de conos (células receptoras de color).
- Definen el modelo CMYK.

# Propiedades del color

Matiz, Saturación y Brillo.

Estas 3 propiedades combinadas entre sí son capaces de sintetizar toda la gama de colores existente por un camino diferente a la combinación de colores primarios aditivos (RGB) y son la base de los modelos HSL y HSV.

## Matiz

También conocido como tono o tonalidad, (en inglés *Hue*), ubica una posición de un color dentro del círculo cromático, por lo que depende de las longitudes de onda.

<div align="center">
    <img src="img/Hue.png" alt="" width="50%" align="center" />
</div>


## Saturación

Es el colorido o pureza y depende de la distancia entre determinado color y la escala de grises; con la mayor saturación un color será más vivo, puro o colorido. En palabras simples, se refiere a la intensidad, es decir, si el color es más sutil o más fuerte.

<div align="center">
    <img src="img/Saturation.png" alt="" width="50%" align="center" />
</div>

## Brillo

Corresponde a la claridad u oscuridad del color, de acuerdo a la luminosidad o al valor. El mínimo valor (cero) siempre es el negro y el máximo depende del modelo (en HSL el 100% es blanco y en HSV 100% es vivo o claro).

# Píxel

- Unidad mínima en color que forma parte de una imagen digital.

- Cada punto representado en la imagen contiene información del color, representada en canales separados (componentes primarios).

- Existen diferentes modelos para representar color: RGB, CMYK, LAB, etc.

# RGB

En el modelo RGB cada pixel se representa como un vector de 3 componentes (Red, Green, Blue) con valores entre 0 y 255.

# CMYK

El modelo CMYK (**C**yan, **M**agenta, **Y**ellow-amarillo y **K**ey-negro) es sustractivo. Por lo tanto, se basa en la absorción de la luz por un objeto: el color que presenta un objeto corresponde a la parte de la luz que incide sobre este y se refleja no siendo absorbida por el objeto, en este caso el papel blanco.

# Modelos HSL y HSV

Son modelos de síntesis aditiva basados en las propiedades del color y sus códigos son coordenadas cilíndricas. Los parámetros son H=matiz o tono (del inglés *hue*), S=saturación (*saturation*), V=valor (*value*) y L=luminosidad (*lightness*). Se pueden representar geométricamente mediante conos, cilindros o cubos.

<div align="center">
    <img src="img/Hsl-hsv_models.png" alt="" width="60%" align="center" />
</div>

## HSV

- Matiz (**H**ue): se refiere a la frecuencia dominante del color dentro del espectro visible, es decir la percepción de un tipo de color y va entre 0° a 360°.
- Saturación (**S**aturation): se refiere a la cantidad del color o a la «pureza» de éste. Va de un color «claro» (0%) a un color más vivo (100%)
- Valor (**V**alue): es la intensidad de luz de un color. Dicho de otra manera, es la cantidad de negro (0%) o de blanco (100%) que posee un color.

# Contraste

El contraste se define como la diferencia relativa en la intensidad entre un punto de una imagen.

<div align="center">
    <img src="img/contraste.jpg" alt="" width="70%" align="center" />
</div>

Cuanto mayor es la distancia en una imagen entre el tono más claro y el más oscuro, mayor contraste tendrá esa foto.
