# Imágenes en 3 canales de color

- Una imagen a color es una imagen digital que incluye información de color para cada píxel.

- En el modelo RGB, diremos que una imagen a color es una matriz $A \in \mathcal{R}^{m \times n \times c} $, donde $m$ representa el alto de la imagen (cantidad de filas), $n$ el ancho de la imagen (cantidad de columnas) y $c=3$ son los canales rojo (R), verde (G) y azul (B).

- La imagen a color está formada por 3 matrices de tamaño $m \times n$. Estas matrices son $A_r$ (canal rojo), $A_g$ (canal verde) y $A_b$ (canal azul). Entonces, $A=\{A_r,A_g,A_b\}$.

- El píxel $A(i,j,r)$ representa la posición $(i,j)$ de la imagen en el canal $r$.

## Visualización de cada canal por separado

```python
def separar_canales(imagen):
    R = imagen[:,:,0]
    G = imagen[:,:,1]
    B = imagen[:,:,2]
    fig, axs = plt.subplots(nrows=1, ncols=4, figsize=(15, 5))
    plt.suptitle("Los 3 canales de color de la imagen", size=16).set_y(0.85)
    axs[0].imshow(imagen)
    axs[0].set_title("Imagen original (3 canales)")
    axs[1].imshow(R, cmap="gray")
    axs[1].set_title("Canal rojo")
    axs[2].imshow(G, cmap="gray")
    axs[2].set_title("Canal verde")
    axs[3].imshow(B, cmap="gray")
    axs[3].set_title("Canal azul")
    plt.show()
```

```python
A = imread('img/rgb.png')
plt.imshow(A)
plt.show()
```

## Visualización de imágenes con 3 canales, pero solo 1 activo

```python
def desactivar_canales(imagen):
    canales = ['rojo','verde','azul']
    plt.figure(figsize=(15, 3))
    plt.suptitle("Imágenes con los 3 canales de color, pero solo uno activo.", size=16).set_y(1.1)
    plt.subplot(1, 4, 1)    
    plt.imshow(imagen)
    plt.title("Imagen original (3 canales)")
    for i,canal in zip(range(3),canales):
        Z = np.zeros(imagen.shape, dtype = int)
        plt.subplot(1, 4, i+2)
        plt.title(f"Solo canal {canal} activo")
        Z[:,:,i] = imagen[:,:,i]
        plt.imshow(Z)
    plt.show()
```

# Blanco y Negro

¿Cómo convertir una imagen en color a blanco y negro?

## ¿ Promediar R + G + B ?
Un enfoque simple podría ser promediar los tres canales de color.

```python
def promediar(imagen):
    imgAvg = (imagen[:,:,0]+imagen[:,:,1]+imagen[:,:,2])/3
    plt.imshow(imgAvg, cmap='gray')
    plt.show()
```

```python
promediar(mario)
```

```python
promediar(raccoon)
```

## Ponderar cada color
Esta no es la forma más adecuada porque debemos tener en cuenta la contribución perceptiva que tiene cada color a la percepción general de la luz.

Los conos no están igualmente distribuidos en la retina y no son iguales en número. Por lo tanto, para obtener una conversión más apropiada deberíamos usar el concepto de Luminancia (Luma).

### Luma

En una transmisión de vídeo, luma es la componente que codifica la información de luminosidad de la imagen. En términos generales, es algo muy similar a la versión en blanco y negro de la imagen original.

La recomendación para HDTV de 2015 es:

$$ Luminance = 0.2126 * R + 0.7152 * G + 0.0722 * B $$

```python
def luma(imagen):
    imgLum=(imagen[:,:,0]*.2126+imagen[:,:,1]*.7152+imagen[:,:,2]*.0722)
    plt.imshow(imgLum, cmap='gray')
    plt.show()
```

```python
luma(mario)
```

```python
luma(raccoon)
```

# Aumentar el brillo

¿Cómo aumentar el brillo?

## Factor multiplicativo

Multiplicar todos los pixeles por algún valor.

```python
def multiplicativa(imagen,factor):
    i=imagen.copy()
    i*=factor
    i[i>255] = 255
    return i
```

```python
def brillo(imagen, forma, factor):
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(13, 5))
    plt.suptitle(f"Aumento del brillo de forma {forma.__name__} en la imagen", size=16).set_y(0.95)
    axs[0].imshow(imagen)
    axs[0].set_title("Imagen original")
    axs[1].imshow(forma(imagen,factor))
    axs[1].set_title(f"Imagen con brillo aumentado x {factor}")
    plt.show()
```

```python
rubik = imread('img/rubik.png')
```

```python
plt.imshow(rubik)
plt.show()
```

```python
brillo(rubik,multiplicativa,2)
```

```python
brillo(raccoon,multiplicativa,3)
```

## Cuidado con el formato

Cuando una imagen está en formato `dtype=uint8` sus valores va en la escala entre 0 y 255. Por lo tanto, si un valor supera el 255 se reinicia; por ejemplo, 165+100=9. En estos casos se debe tranformar a `dtype=uint64`, para poder realizar operaciones de aumento de píxeles libremente y luego acotar a 255 si se excede el límite.

```python
rubik = rubik.astype(np.int64)
raccoon = raccoon.astype(np.int64)
```

```python
brillo(rubik,multiplicativa,2)
```

```python
brillo(raccoon,multiplicativa,3)
```

## Factor sumativo

Sumar todos los pixeles por algún valor.

```python
def sumativa(imagen, factor):
    i=imagen.copy()
    i[:,:,0] = i[:,:,0] + factor
    i[:,:,1] = i[:,:,1] + factor
    i[:,:,2] = i[:,:,2] + factor
    i[i>255] = 255
    return i
```

```python
brillo(raccoon,sumativa,70)
```

```python
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15, 5))

factor = [0, 70, 200]

for i in range(3):
    array = sumativa(raccoon, factor[i])
    ax[i].imshow(array)
```

# Pixelar

¿Cómo lograr el efecto Pixelado?

Restringir la cantidad de pixeles para bajar la calidad de la imagen.

```python
def pixelar(imagen,factor):
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(13, 5))
    plt.suptitle(f"Imagen pixelada", size=16).set_y(0.95)
    axs[0].imshow(imagen)
    axs[0].set_title("Imagen original")
    axs[1].imshow(imagen[::factor,::factor])
    axs[1].set_title(f"Imagen con efecto pixelado x {factor}")
    plt.show()
```

```python
pixelar(rubik,5)
```

```python
pixelar(raccoon,50)
```

# Efecto Negativo

¿Cómo lograr el efecto de negativo?

Invertir el tono, restándole a 255 el valor del píxel.

```python
def negativo(imagen):
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(13, 5))
    plt.suptitle(f"Negativo de la imagen", size=16).set_y(0.95)
    axs[0].imshow(imagen)
    axs[0].set_title("Imagen original")
    N = np.full(imagen.shape,255) - imagen
    axs[1].imshow(N)
    axs[1].set_title(f"Imagen con efecto negativo")
    plt.show() 
```

```python
negativo(rubik)
```

```python
negativo(mario)
```

```python
negativo(raccoon)
```

# Contraste

¿Cómo aumentar el contraste a una imagen?

Definir un umbral, donde todos los píxeles con valores menores al umbral van al  mínimo (0) y los mayores o iguales van al máximo (255).

```python
def contraste(imagen,umbral):
    i=imagen.copy()
    i[i<umbral]=0
    i[i>=umbral]=255
    return i
```

```python
def contrastar(imagen,umbral):
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(13, 5))
    plt.suptitle(f"Aumento del contraste en la imagen", size=16).set_y(1)
    axs[0].imshow(imagen)
    axs[0].set_title("Imagen original")
    axs[1].imshow(contraste(imagen,umbral))
    axs[1].set_title(f"Imagen con contraste aumentado x {umbral}")
    plt.show()
```

```python
contrastar(rubik,125)
```

```python
contrastar(raccoon,125)
```

```python
contrastar(mario,150)
```

```python
colores = plt.imread('img/colores.jpeg')
contrastar(colores,125)
```

# Ajustar el Matiz

```python
def matiz(imagen, canal, correccion):
    i=imagen.copy()
    i[:,:,canal]=i[:,:,canal]*correccion
    i[i>255]=255
    return i
```

```python
def ajustar_matiz(imagen, canal, correccion):
    d = {'rojo' : 0, 'verde' : 1, 'azul' : 2}
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(13, 5))
    plt.suptitle(f"Ajuste del tono en la imagen", size=16).set_y(0.95)
    axs[0].imshow(imagen)
    axs[0].set_title("Imagen original")
    axs[1].imshow(matiz(imagen, d[canal], correccion))
    axs[1].set_title(f"Imagen con el canal {canal} ajustado en {correccion}")
    plt.show()
```

```python
ajustar_matiz(rubik, 'verde', 1.8)
```

```python
d = {'rojo' : 0, 'verde' : 1, 'azul' : 2}
f, axs = plt.subplots(3,3,figsize=(12,8))
for canal in d:
    fila = 0
    for correccion in (2.5,.75,0.2):
        axs[d[canal],fila].imshow(matiz(rubik,d[canal],correccion))
        axs[d[canal],fila].axis('off')
        axs[d[canal],fila].set_title(f"Canal {canal} ajustado en {correccion}")
        fila += 1
f.tight_layout()
```
