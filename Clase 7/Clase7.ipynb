# Aprendizaje Supervisado

## Regresión 

- La regresión predice valores continuos (números).
- Por ejemplo: Cantidad de mm de lluvia mañana.

## Clasificación

- La clasificación predice valores discretos (etiquetas/clases).
- Por ejemplo: llueve o no llueve mañana (binaria); perro, gato, conejo (multiclase).

# Aprendizaje Supervisado

## Conjunto de Entrenamiento (Training set)

Es el conjunto de datos que contiene las etiquetas de cada registro y permite al modelo aprender.

## Conjunto de Validación (Validation set)

Se utiliza cuando se debe ajustar ciertos parámetros del algoritmo.

## Conjunto de Pruebas (Testing set)

Es un conjunto de datos nuevo y desconocido para el modelo y se utiliza para evaluar el rendimiento del modelo una vez entrenado. Este conjunto nunca debe ser utilizado en la etapa de entrenamiento porque se pierde generalidad.

# Generalización

Capacidad de un modelo de predecir con precisión no sólo los datos con los que fue entrenado, sino que también los datos que
no le habían sido presentados antes.

# Función Objetivo

- Es una función matemática que mide la pérdida (costo) o la ganancia del modelo. En el primer caso debe ser minimizada y en el segundo caso maximizada.

- En regresión se usa como función objetivo el Mean Absolute Error o el Mean Squared Error.

- En clasificación binaria se usa la Binary Cross-Entropy (Log Loss) o la Hinge Loss y en clasificación multi-clase la más común es la Categorical Cross-Entropy (Softmax Loss).

# Optimización

- Proceso para encontrar o aproximarse al óptimo de una función objetivo.
- El método más común es la gradiente descendente.
- Se actualizan los parámetros de forma iterativa para ir disminuyendo el error en cada "época".

# Gradiente Descendente

Calcula, el gradiente de la función objetivo en cierto punto y elige la dirección que tiene "mayor inclinación" (dirección opuesta del gradiente).

$$ \text{w}_{i+1} \leftarrow \text{w}_i - \alpha \nabla_\text{w} J(\text{w}) $$

# Entrenamiento
Es el proceso mediante el cual el optimizador minimiza/maximiza la función objetivo y al hacerlo, encuentra los parámetros del modelo que mejor se **ajustan** a la tarea que queremos.

# Underfitting (sub-ajuste)

El modelo no aprendió lo suficiente y la calidad predictiva es deficiente. Puede que el modelo sea muy simple para el problema a resolver, no se ajustaron bien los parámetros o necesita mayor cantidad de iteraciones.

<div align="center">
    <img src="img/underfitting.png" alt="" width="50%" align="center" />
</div>

# Overfitting (sobre-ajuste)
Fenómeno en el cual nuestro modelo aprendió demasiado bien el conjunto de datos de entrenamiento. Como consecuencia de esto, el modelo generó reglas internas que se apegan demasiado a estos datos (training set) y al evaluarlo en datos que nunca antes ha visto (testing set), se comporta mal.

## Conceptos básicos

### Neuronas

Elemento básico de una red neuronal, recibe un cierto input y al procesarlo genera un output de acuerdo a una función de activación.

<div align="center">
    <img src="img/neuron.png" alt="" width="30%" align="center" />
</div>

### Función de activación

Se encarga de entregar una salida a partir de un valor de entrada, calculando el resultado o "activación" de la neurona.

<div align="center">
    <img src="img/activations.png" alt="" width="100%" align="center" />
</div>    

### Capas

Serie de neuronas que ocupan una determinada posición paralela en la secuencialidad de una red.
    
<div align="center">
    <img src="img/layers.png" alt="" width="30%" align="center" />
</div>    

- **Input (entrada):** Es la capa que representa el vector de entrada de datos a la red neuronal.

- **Hidden (oculta):** Son todas las capas intermedias entre la entrada y la salida.

- **Output (salida):** Es la capa de salida que representa el resultado entregado por la red neuronal.
